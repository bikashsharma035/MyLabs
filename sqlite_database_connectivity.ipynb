```python
import sqlite3 as sq #alias/ref
```

```python
con = sq.connect("DB/pythonDB2.db")
```

```python
cur = con.cursor()
```

```python
cur.execute('create table employee(empid int primary key,empname text,location text,salary double(7,2))')
con.commit()
```

```python
cur.execute("insert into employee values (11,'Neha','Kolkata',67000.00)")
cur.execute("insert into employee values (12,'Aparna','Gurugram',90000.00)")
cur.execute("insert into employee values (13,'Prakash','Bangalore',100000.00)")
cur.execute("insert into employee values (14,'John','Pune',76000.00)")
cur.execute("insert into employee values (15,'Dinesh','Hyderabad',87000.00)")
con.commit()
```

```python
cur.execute("select * from employee")
data=cur.fetchall()
print(data)
```

```python
cur.execute("update employee set empname='Nisha',location='Pune' where empid =11")
con.commit()
cur.execute("select * from employee")
data=cur.fetchone()
#data=cur.fetchall()
#data=cur.fetchmany(5)
print(data)
```

```python
cur.execute("delete from employee where empid=14")
con.commit()
cur.execute("select * from employee")
data=cur.fetchall()
print(data)
```
