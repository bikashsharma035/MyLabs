```python
import  numpy as np
myarr=np.array([[1,2,3,4],
                [1,2,5,6],
                [1,3,4,5]])#3X4
print(myarr)
```

```python
print(np.max(myarr)) #finding max value from matrix,#6
print(np.min(myarr)) #finding min value from matrix,#1
print(np.sum(myarr)) #calculating sum of all the values of matrix #37
print(np.max(myarr,axis=0)) #finding max value column wise #[1,3,5,6]
print(np.min(myarr,axis=0)) #fining min value column wise  #[1,2,3,4]
print(np.sum(myarr,axis=0)) #calculating sum column wise # [3,7,12,15]
print(np.max(myarr,axis=1)) #finding max value row wise #[4,6,5]
print(np.min(myarr,axis=1)) #fining min value row wise #[1,1,1]
print(np.sum(myarr,axis=1)) #calculating sum row wise #[10,14,13]
print(myarr.shape) #no. of rows and columns #3X4
print(myarr.size) #total no. of elements #12
```

```python
print(myarr.flatten()) #combines all the elements of array
```

```python
x = np.array([[1,2],[3,4]]) #x = np.array([1,2,3,4]
y = np.array([[3,5],[1,4]])
#array x contains two lists: [1, 2] and [3, 4].
#The first list [1, 2] becomes the first row of the matrix.
#The second list [3, 4] becomes the second row of the matrix.

print(x)
print(y)
```

```python
#generate random numbers from normal distribution with mean 0 and standard deviation 1
rnd = np.random.normal(0,1,size=(3,3))
#np.random.normal() is a function from the NumPy random module that generates random numbers from a normal (Gaussian) distribution
#loc: The mean (center) of the distribution. In this case, it's 0. This means the distribution is centered around 0.
#scale: The standard deviation (spread or width) of the distribution. Here, it's 1. This means the spread is based on a standard normal distribution (i.e., a normal distribution with a mean of 0 and a standard deviation of 1).
#size: The shape of the output array. Here, it's (3, 3), meaning a 3x3 matrix (a 2D array with 3 rows and 3 columns) will be created.
rnd
```

```python
np.random.seed(4)
#np.random.seed() is used to set the seed for the random number generator in NumPy
#The seed is a starting point for generating a sequence of random numbers. 
#By setting the seed to a fixed number (in this case, 4), you ensure that the random numbers generated will be reproducible. This means every time you run the code with the same seed (4), you will get the same sequence of random numbers.
rnd = np.random.rand(3)
rnd
```

```python
rand_vec = np.random.randn(15)
#np.random.randn() generates random numbers from the standard normal distribution, meaning the numbers follow a normal distribution with a mean of 0 and a standard deviation of 1.
#The argument 15 specifies that we want to generate 15 random values from this normal distribution.
print(rand_vec)
##The resulting rand_vec is a 1D NumPy array of 15 random floating-point values drawn from the standard normal distribution. These values can be both positive and negative.
print()
print(rand_vec>0)#rand_vec > 0 creates a boolean mask or boolean array.False if <0 and True if >0
```
