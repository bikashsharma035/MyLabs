```python
import pandas as pd
```

```python
#dataframe using list
list1=[33,55,66,22,12,44,67]
df=pd.DataFrame(list1)
print(df)
```

```python
Pandas automatically treats the list as a single column, and each item in the list 
corresponds to a row in that column. By default, it assigns an index (0, 1, 2, ...) to each row.
```

```python
#dataframe using dictionary of list
details = { 
    'Name' : ['Ankit', 'Aishwarya', 'Shaurya', 'Shivangi'], 
    'Age' : [23, 21, 22, 21], 
    'University' : ['BHU', 'JNU', 'DU', 'BHU'], 
} 
df=pd.DataFrame(details)
print(df)
```

```python
pd.DataFrame(details) creates a pandas DataFrame using the details dictionary.
The dictionary keys (i.e., 'Name', 'Age', and 'University') are used as column labels, 
and the values in each list are populated as rows under the corresponding column.
```

Dataframe - tabular format - rows and columns

```python
# Python code demonstrate how to create
# Pandas DataFrame by lists of dicts

# Initialize data to lists.
data = [{'a': 1, 'b': 2, 'c': 3},
 {'a': 10, 'b': 20, 'c': 30},
  {'a': 100, 'b': 200, 'c': 300}]

# Creates DataFrame.
#df=pd.DataFrame(data)
df = pd.DataFrame(data, index=['First Row','Second Row','Third Row'] )

# Print the data
df
```

```python
Columns: 'a', 'b', and 'c' come from the keys in each dictionary.
Rows: The values from the dictionaries are placed under the appropriate column.
Index: The row labels First Row, Second Row, and Third Row replace the default numerical index.
```

```python
rd = df.loc['Second Row']
rd
```

```python
.loc indexer in pandas is used to access a group of rows and columns by labels (i.e., row and column names).
```

```python
dict = {"Name": ["Anush Sharma", "Balareddy", "Chintu", "Eshwari", "Fathima"],
       "Occupation": ["Rocket Scientist", "Rocket Scientist", "Lawyer", "Data Scientist", "Rocket Scientist"],
       "ID": [1, 2, 3, 4, 5]}

df = pd.DataFrame(dict)#converts the dictionary into a pandas DataFrame
print(df)

print(df['Occupation'] == 'Rocket Scientist')
```

```python
This would return only the rows where the "Occupation" is 'Rocket Scientist'.
It creates a boolean Series (a series of True or False values) where each value represents whether 
the occupation in the corresponding row is equal to 'Rocket Scientist'.
```

```python
myjson='[{"id":"101","Name":"Neha","age":"33"},{"id":"102","Name":"Nootan","age":"30"}]'
df3 =pd.read_json(myjson)
print(df3)
```

```python
pd.read_json() function is a pandas method that reads a JSON-formatted string and converts it into a
pandas DataFrame.
JSON Keys (e.g., id, Name, age) will become the column labels of the DataFrame.
JSON Objects will become the rows of the DataFrame.
```

```python
import json
from pandas import json_normalize #json module to load the json data
myjson='[{"id":"101","Name":"Neha","age":"33"},{"id":"102","Name":"Nootan","age":"30"}]'#string containing JSON data
dict1 = json.loads(myjson)
df3 =json_normalize(dict1)# convert the list of dictionaries (dict1) into a pandas DataFrame.
print(df3)
```

```python
df2=pd.read_csv('dataset/missing_value.csv',na_values=['not','na'])
# any occurrence of the values 'not' or 'na' in the CSV file will be treated as missing data (represented as NaN in the DataFrame).
df2
```

```python
df2.isnull().sum()
#True is treated as 1(having NaN values) and False as 0(No no NaN values) , so summing gives the total number of NaN values in each column
```

```python
df2['ranking'].unique()#returns an array of unique values in a Series.
```

```python
df2.dropna(axis=0,inplace=True)#remove missing data from a DataFrame or Series
#axis=0 specifies that rows should be dropped. In pandas,axis=0 refers to rows (the default behavior).
#axis=1 would refer to columns.
df2
```

```python
df5 = df2['product'].drop_duplicates()#removes duplicate values in a Series or DataFrame.
df5
#shampoo: This value appears multiple times in the product column, but only the first occurrence (index 0) is kept.
#sugar: This value appears multiple times, but only the first occurrence (index 7) is kept.
#salt: This value appears multiple times, but only the first occurrence (index 11) is kept.
```

```python
df2.rename(columns={'product':'items'},inplace=True)#rename columns or rows in a DataFrame
#inplace=True means that the renaming operation will modify the original DataFrame (df2)
df2
```

```python
filtered_df = df2['items']=='shampoo' #create a boolean mask or filter that identifies the rows in the df2
filtered_df
#True is returned for rows where the items column is equal to 'shampoo'.
#False is returned for rows where the items column is not equal to 'shampoo'.
```

```python
filtered_df = (df2['items']=='shampoo') & (df2['name']=='A')#applies two conditions simultaneously
#True if both the conditions are true as this is AND operator
filtered_df
```
