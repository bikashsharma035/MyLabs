```python
## Pandas
#Series
import pandas as pd
#This imports the pandas library and assigns it the alias pd
import numpy as np
#This imports the numpy library with the alias np
s1=pd.Series()#Empty
#This creates an empty Series.
print(s1)# Empty Series
s2=pd.Series([2,10,3,6,'Hello'])
#This creates a Series with the given data. The data is provided as a list [2, 10, 3, 6, 'Hello'], which will be converted into a pandas Series
# The default index for each element will be integers starting from 0.
print(s2)
#dtype: object indicates that the data type of the Series is object
s3=pd.Series({1:"Id",2:"Name"})
#This creates a Series from a dictionary. The dictionary {1: "Id", 2: "Name"} is passed, which means the keys (1, 2) become the index labels, and the corresponding values ("Id", "Name") become the data of the Series.
print(s3)
print(s3.index)
#This prints the index of s3. The index is [1, 2].
s4=pd.Series(np.array([2,5,8,12]))
#This creates a Series from a NumPy array. The array [2, 5, 8, 12] is passed to create a pandas Series. Again, pandas will assign default integer indices starting from 0.
print(s4)
```

```python
## Pandas
#Series

import pandas as pd
import numpy as np
s1=pd.Series()#Empty
print(s1)# Empty Series
s2=pd.Series([2,10,3,6,'Hello'])
print(s2)
s3=pd.Series({1:"Id",2:"Name"})
#print(s3)#Data type: object (because the Series contains strings).
print(s3.index)#This prints the index of s3. The index is [1, 2].
print(s3.ndim)#ndim returns the number of dimensions in the Series. A pandas Series is a one-dimensional structure, so it will always return 1
print(s3.shape)#shape returns a tuple representing the dimensions of the Series. For a one-dimensional Series, it will return a tuple with a single value representing the number of elements in the Series
print(s3.dtypes)#dtypes returns the data type of the elements in the Series
print(s3.hasnans)#hasnans checks whether there are any NaN (Not a Number) values in the Series
s4=pd.Series({1:"Id",2:"Name",3:np.nan})
print(s4)
print(s4.hasnans)
print(s1.empty)#empty checks whether the Series s1 is empty (i.e., it contains no elements).
print(s1.dtypes)# Since s1 is empty and contains no data, pandas assigns a default data type of float64 for empty Series
```

```python
import numpy as np
import pandas as pd
x1=np.random.rand(2,3)
df1=pd.DataFrame(x1,columns=['Id','Name','Address'],index=['Row1','Row2'])
print(df1)
print("")
print("Display col")
print(df1['Id'])
 
#display rows:
# loc-->label
# iloc-->index num
 
print("")
print("Display Rows using loc")
print(df1.loc['Row1']) #.loc[] selects rows by label.'Row1' is the index label.
 
print("")
print("Display Rows using iloc")
print(df1.iloc[1])#.iloc[] selects rows by position/index number.1 refers to the second row (Python is zero-indexed).
```

```python
#Slicing for Pandas
import pandas as pd
ds5=pd.read_csv(r'C:\Users\bikash.a.sharma\OneDrive - Accenture\Python BootCamp_6-10thOct_2025\Day 4\Pandas Demos\student.csv')
#print(ds5)
#iloc: integer based indexing
r1=ds5.iloc[0:5] #accept row index 0(1st row) to till Row Index 4(5th row)
print(r1)
r2=ds5.iloc[1:6]#acceps Row Index 1(2nd row) to till Row Index 5(6th row)
print(r2)
r3=ds5.iloc[5,3] #RowIndex 5(6th row) and ColIndex3(4th Col)
print(r3)
print(ds5.iloc[:,0:2])#1st param is row and 2nd param  is col
r4=ds5.iloc[11,4]# Row Index 11(12th row) & Col Index 4(5th Col)
print(r4)
#loc[] is used for label based indexing (row/col labels)
#start and end labels are inclusive
#df.loc[1:3] will include 2nd row and 4th col
#iloc[]: end pos will be exclusive
```
